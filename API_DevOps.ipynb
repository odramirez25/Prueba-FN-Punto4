###2. Creación de la API REST con sus respectivos endpoints.
Para la creación de la API se utilizó la librería FastApi. Se decidió crear 4 contratos (endpoints) que representan las operaciones que los cliente pueden realizar con la API, las operaciones son **carga de departamentos**, **carga de puestos de trabajos**, **carga de empleados** y **consulta de información**.

####Endpoints para carga de información
#####Enpoints 1 y 2
En estos endpoints se realiza la carga de los datos de las tablas para los datos de los departamentos y puestos de trabajo a partir de sus dataframe correspondientes creados en el script de generación de datos.

#####Endpoint 3
En este endpoint se realiza la carga de los datos de la tabla para los datos de los empleados a partir del archivo CSV generado en el script para crear archivos CSV y Parquet con los datos generados en el script de generación de datos. Adicionalmente, para este caso se usa la declaración UploadFile = File(...). Esta declaración habilita el protocolo multipart/form-data, el cual permite que el cliente transmita un archivo binario o de texto masivo por la red sin cargarlo completo en la memoria RAM del servidor de un solo golpe y usa streaming con la función await file.read(). También se realiza la validación de que el archivo sea de tipo CSV.

Para estos 3 endpoints se utilizaron los decoradores @app.post("/batch/departments"), @app.post("/batch/jobs"), @app.post(/upload/employees-csv) y verbo HTTP POST para mapear las URL y se maneja el código respuesta HTTP, HTTP 201 CREATED.

Para poder hacerla es necesario hacer join con la tabla de empleados, un group by para agrupar el resultado por cada departamento y finalmente ordena el resultado de manera descendente por el total de empleados.

Para este endpoint se utilizó el decorador @app.get("/reports/department-summary") y verbo HTTP GET para mapear las URL y se maneja el código respuesta HTTP por defecto, HTTP 200 OK.

Para todos los endpoints se realiza un manejo de sesiones de la siguiente manera:

```
db = SessionLocal() --> 1. Abre una conexión
try:
    db.commit()  --> 2. Confirma los cambios si todo sale bien
except Exception as e:
    db.rollback() --> 3. Revierte la transacción si hay error (Mantiene la integridad)
finally:
    db.close() --> 4. Cierra la sesión (Evita que queden conexiones abiertas)
```

````python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Se fija semilla para reproducibilidad
np.random.seed(42)

# ==========================================
# 1. TABLA: Departamentos (departments)
# ==========================================
departamentos_data = {
    "department_id": [1, 2, 3, 4, 5],
    "department_name": ["Tecnología", "Recursos Humanos", "Finanzas", "Ventas", "Operaciones"]
}
df_departamentos = pd.DataFrame(departamentos_data)

# ==========================================
# 2. TABLA: Puestos de Trabajo (jobs)
# ==========================================
puestos_data = {
    "job_id": [101, 102, 103, 104, 105, 106],
    "job_title": [
        "Ingeniero de Datos",
        "Arquitecto Cloud",
        "Analista de RRHH",
        "Contador Senior",
        "Ejecutivo de Ventas",
        "Gerente de Operaciones"
    ],
    "min_salary": [4000, 6000, 2500, 3000, 2000, 5000],
    "max_salary": [8000, 12000, 4500, 5500, 6000, 9000]
}
df_puestos = pd.DataFrame(puestos_data)

# ==========================================
# 3. TABLA: Empleados (employees)
# ==========================================
num_empleados = 1000

# Nombres y apellidos base para simular
nombres = ["Carlos", "Ana", "Luis", "Maria", "Javier", "Sofia", "Oscar", "Laura", "Andres", "Valentina"]
apellidos = ["Gomez", "Rodriguez", "Lopez", "Martinez", "Garcia", "Perez", "Torres", "Ramirez", "Sanchez", "Vargas"]

# Generación aleatoria de atributos de empleados
emp_id = np.arange(10001, 10001 + num_empleados)
emp_nombres = np.random.choice(nombres, num_empleados)
emp_apellidos = np.random.choice(apellidos, num_empleados)

# Generación de correos electrónicos únicos
emp_emails = [
    f"{nom.lower()}.{ape.lower()}{i}@empresa.com"
    for i, (nom, ape) in enumerate(zip(emp_nombres, emp_apellidos), start=1)
]

# Asignación de departamentos y puestos aleatorios
emp_dept_ids = np.random.choice(df_departamentos["department_id"], num_empleados)
emp_job_ids = np.random.choice(df_puestos["job_id"], num_empleados)

# Fechas de contratación aleatorias entre 2018 y 2025
start_date = datetime(2018, 1, 1)
end_date = datetime(2025, 12, 31)
time_between_dates = end_date - start_date
days_between_dates = time_between_dates.days

random_number_of_days = np.random.randint(0, days_between_dates, num_empleados)
emp_hire_dates = [start_date + timedelta(days=int(d)) for d in random_number_of_days]

# Generación de salarios dentro del rango de cada puesto
job_min_map = dict(zip(df_puestos["job_id"], df_puestos["min_salary"]))
job_max_map = dict(zip(df_puestos["job_id"], df_puestos["max_salary"]))

emp_salarios = [
    np.round(np.random.uniform(job_min_map[jid], job_max_map[jid]), 2)
    for jid in emp_job_ids
]

# Crear el DataFrame final de empleados
df_empleados = pd.DataFrame({
    "employee_id": emp_id,
    "first_name": emp_nombres,
    "last_name": emp_apellidos,
    "email": emp_emails,
    "hire_date": emp_hire_dates,
    "department_id": emp_dept_ids,
    "job_id": emp_job_ids,
    "salary": emp_salarios
})

# ==========================================
# Visualización previa
# ==========================================
print("--- DEPARTAMENTOS ---")
print(df_departamentos.head(2))
print("\n--- PUESTOS ---")
print(df_puestos.head(2))
print("\n--- EMPLEADOS (Muestra) ---")
print(df_empleados.head(3))
````

````python
import io
from pathlib import Path
from typing import List
import pandas as pd
from pydantic import BaseModel
from sqlalchemy import (
    Column,
    Float,
    ForeignKey,
    Integer,
    String,
    create_engine,
    text,
)
from sqlalchemy.orm import declarative_base, sessionmaker

# 1. Configuración de Base de Datos Relacional Portable (SQLite)
DB_PATH = Path.cwd() / "hr_company.db"
DATABASE_URL = f"sqlite:///{DB_PATH}"

engine = create_engine(
    DATABASE_URL, connect_args={"check_same_thread": False}
)
SessionLocal = sessionmaker(autocommit=False, autoflush=False, bind=engine)
Base = declarative_base()


# Se definen los modelos ORM para poder crear las tablas
class DepartmentModel(Base):
    __tablename__ = "departments"
    department_id = Column(Integer, primary_key=True)
    department_name = Column(String, nullable=False)


class JobModel(Base):
    __tablename__ = "jobs"
    job_id = Column(Integer, primary_key=True)
    job_title = Column(String, nullable=False)
    min_salary = Column(Float)
    max_salary = Column(Float)


class EmployeeModel(Base):
    __tablename__ = "employees"
    employee_id = Column(Integer, primary_key=True)
    first_name = Column(String)
    last_name = Column(String)
    email = Column(String, unique=True)
    hire_date = Column(String)
    job_id = Column(Integer, ForeignKey("jobs.job_id"))
    salary = Column(Float)
    department_id = Column(Integer, ForeignKey("departments.department_id"))


# Se crean las estructuras de tablas
Base.metadata.create_all(bind=engine)


# 3. Schemas de Validación (Pydantic)
class DepartmentSchema(BaseModel):
    department_id: int
    department_name: str


class JobSchema(BaseModel):
    job_id: int
    job_title: str
    min_salary: float
    max_salary: float


# 4. Inicialización de FastAPI
from fastapi import FastAPI, File, HTTPException, UploadFile, status
from fastapi.testclient import TestClient

app = FastAPI(
    title="HR Data Platform API",
    description="API REST para Ingesta y Analítica de RRHH",
)


# Endpoint 1: Ingesta por lote (Batch) de Departamentos
@app.post("/batch/departments", status_code=status.HTTP_201_CREATED)
def insert_departments_batch(departments: List[DepartmentSchema]):
    db = SessionLocal()
    try:
        for dept in departments:
            db_dept = DepartmentModel(**dept.model_dump())
            db.merge(db_dept)
        db.commit()
        return {
            "status": "success",
            "inserted_records": len(departments),
            "message": "Lote de departamentos procesado exitosamente.",
        }
    except Exception as e:
        db.rollback()
        raise HTTPException(status_code=400, detail=str(e))
    finally:
        db.close()


# Endpoint 2: Ingesta por lote (Batch) de Puestos de Trabajo
@app.post("/batch/jobs", status_code=status.HTTP_201_CREATED)
def insert_jobs_batch(jobs: List[JobSchema]):
    db = SessionLocal()
    try:
        for job in jobs:
            db_job = JobModel(**job.model_dump())
            db.merge(db_job)
        db.commit()
        return {
            "status": "success",
            "inserted_records": len(jobs),
            "message": "Lote de puestos procesado exitosamente.",
        }
    except Exception as e:
        db.rollback()
        raise HTTPException(status_code=400, detail=str(e))
    finally:
        db.close()


# Endpoint 3: Carga de Archivos (CSV Upload) para Empleados
@app.post("/upload/employees-csv", status_code=status.HTTP_201_CREATED)
async def upload_employees_csv(file: UploadFile = File(...)):
    if not file.filename.endswith(".csv"):
        raise HTTPException(
            status_code=400, detail="El archivo debe ser en formato CSV"
        )

    contents = await file.read()
    df_upload = pd.read_csv(io.BytesIO(contents))

    db = SessionLocal()
    try:
        records = df_upload.to_dict(orient="records")
        for rec in records:
            emp = EmployeeModel(**rec)
            db.merge(emp)
        db.commit()
        return {
            "status": "success",
            "filename": file.filename,
            "total_ingested": len(records),
            "message": "Archivo CSV procesado e ingestado correctamente en la base de datos SQL.",
        }
    except Exception as e:
        db.rollback()
        raise HTTPException(
            status_code=500, detail=f"Error durante la ingesta: {str(e)}"
        )
    finally:
        db.close()


# Endpoint 4: Reporte / Vista Analítica SQL (JOIN + GROUP BY)
@app.get("/reports/department-summary")
def get_department_summary():
    db = SessionLocal()
    try:
        query = text("""
            SELECT
                d.department_name AS departamento,
                COUNT(e.employee_id) AS total_empleados,
                ROUND(SUM(e.salary), 2) AS nomina_total,
                ROUND(AVG(e.salary), 2) AS salario_promedio,
                ROUND(MIN(e.salary), 2) AS salario_minimo,
                ROUND(MAX(e.salary), 2) AS salario_maximo
            FROM departments d
            LEFT JOIN employees e ON d.department_id = e.department_id
            GROUP BY d.department_id, d.department_name
            ORDER BY total_empleados DESC;
        """)
        result = db.execute(query).fetchall()
        report_data = [dict(row._mapping) for row in result]
        return {
            "report_name": "Resumen Ejecutivo por Departamento",
            "data": report_data,
        }
    finally:
        db.close()


# ==========================================
# PRUEBAS DE FUNCIONAMIENTO
# ==========================================
client = TestClient(app)

print("--- 1. Probando Ingesta Batch: Departamentos ---")
dept_payload = df_departamentos.to_dict(orient="records")
res_dept = client.post("/batch/departments", json=dept_payload)
print(res_dept.json())

print("\n--- 2. Probando Ingesta Batch: Puestos ---")
jobs_payload = df_puestos.to_dict(orient="records")
res_jobs = client.post("/batch/jobs", json=jobs_payload)
print(res_jobs.json())

print("\n--- 3. Probando Upload CSV: Empleados (desde data/processed) ---")
path_csv_upload = Path.cwd() / "data" / "processed" / "employees.csv"
with open(path_csv_upload, "rb") as f:
    res_upload = client.post(
        "/upload/employees-csv",
        files={"file": ("employees.csv", f, "text/csv")},
    )
print(res_upload.json())

print("\n--- 4. Probando Vista Analítica (Reporte SQL) ---")
res_report = client.get("/reports/department-summary")
df_reporte = pd.DataFrame(res_report.json()["data"])
display(df_reporte)
````
