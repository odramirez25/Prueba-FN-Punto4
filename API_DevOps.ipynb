## Desafío 1: Generar datos de forma automática
Construya un script que genere de forma automática los datos de: departamentos, puestos de trabajo, y empleados.

Explicación de funcionamiento del script.

Se plantea generar 3 Dataframes para cada tabla requerida con los datos generados.

Para las tablas departamentos y puestos de trabajo se definen listas con algunos datos que referencien la necesidad y a partir de las listas se crean los Dataframes.

Para la tabla Empleados si se llevan a cabo varias operaciones para generar los datos.

*  Se define generar datos de 1000 empleados.
*  Se construye una lista con para el id del empleado y se asignan valores entre 10001 y el número de empleados es decir 11000
*  Se crea la lista de nombres asignando nombres aleatorios de un conjunto de nombres previamente definido.
*  Se crea la lista de apellidos asignando apellidos aleatorios de un conjunto de apellidos previamente definido.
*  Usando las listas de nombres y apellidos se generan una lista de correos electrónicos.
*  Usando los dataframes de departamentos y puestos de trabajo se generan listas aleatorias con los id de departamentos y id de puestos de trabajo
*  Se genera una lista de fechas de contratación de manera aleatoria en un rango entre 2018 y 2025.
*  Usando el dataframe puestos de trabajo se genera una lista de salarios aleatoriamente.
*  Finalmente con las listas de id de empleados, nombres, apellidos, correos electrónicos, fechas de contratación, id de departamentos, id de puestos de trabajo y salarios se crea el dataframe de empleados.

Se realiza un previsualización de los primeros registros de cada dataframe.

###Decisiones técnicas para el desafío 1
1. se fija una semilla para reproducibilidad, para garantizar que siempre se genere el mismo conjunto de datos en cada ejecución del script.
2. Para los correos electrónicos se agrega un valor al usuario para evitar que se generen correos electrónicos repetidos, por ejemplo oscar.gomez1@empresa.com
3. Para la lista de salarios se asegura que el salario asignado se encuentre dentro del rango de salarios del puesto de trabajo, esto se realiza con la combinación de las funciones dict y zip emparejando los id del puesto de salario con los valores mínimos y máximos de los salarios, para finalmente asignar un salario dentro del rango correspondiente según el id del puesto de trabajo
4. Para la generación de los datos se asignaron id para todas los 3 dataframes con la intención de poder tener integridad referencial y en los casos de los dataframes departamentos y puestos de trabajo los id de estos dataframes podrán ser referenciados desde la tabla empleados, esto se hace para que puedan ser utilizados sin inconvenientes para los desafíos posteriores.

## Desafio 2: Guardar los datos simulados en archivos
Guarde los datos simulados en archivos con formato CSV/Parquet.

Explicación de funcionamiento del script.

Usando la librería Path de python se crea el directorio donde van a quedar los archivos, se generan las rutas y nombres para los archivos.

Con las funcionalidades de los dataframes se generan los archivos en formato CSV y Parquet.

Como paso adicional y opcional se realiza una tabla comparativa del espacio usado por los dos archivos, presentando al archivo tipo Parquet como el más eficiente a nivel de espacio debido al uso de compresión.

###Decisiones técnicas para el desafío 2
1. Debido a que el desarrollo se está realizando en Colab de Google, se aseguró que el directorio base se pudiera crear de forma segura en otros entornos también.
2. Se contaba con la opción de agregar un código para realizar una comparación de los tipos de archivo CSV y Parquet, por lo que decidí incluirlo.


```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Se fija semilla para reproducibilidad
np.random.seed(42)

# ==========================================
# 1. TABLA: Departamentos (departments)
# ==========================================
departamentos_data = {
    "department_id": [1, 2, 3, 4, 5],
    "department_name": ["Tecnología", "Recursos Humanos", "Finanzas", "Ventas", "Operaciones"]
}
df_departamentos = pd.DataFrame(departamentos_data)

# ==========================================
# 2. TABLA: Puestos de Trabajo (jobs)
# ==========================================
puestos_data = {
    "job_id": [101, 102, 103, 104, 105, 106],
    "job_title": [
        "Ingeniero de Datos",
        "Arquitecto Cloud",
        "Analista de RRHH",
        "Contador Senior",
        "Ejecutivo de Ventas",
        "Gerente de Operaciones"
    ],
    "min_salary": [4000, 6000, 2500, 3000, 2000, 5000],
    "max_salary": [8000, 12000, 4500, 5500, 6000, 9000]
}
df_puestos = pd.DataFrame(puestos_data)

# ==========================================
# 3. TABLA: Empleados (employees)
# ==========================================
num_empleados = 1000

# Nombres y apellidos base para simular
nombres = ["Carlos", "Ana", "Luis", "Maria", "Javier", "Sofia", "Oscar", "Laura", "Andres", "Valentina"]
apellidos = ["Gomez", "Rodriguez", "Lopez", "Martinez", "Garcia", "Perez", "Torres", "Ramirez", "Sanchez", "Vargas"]

# Generación aleatoria de atributos de empleados
emp_id = np.arange(10001, 10001 + num_empleados)
emp_nombres = np.random.choice(nombres, num_empleados)
emp_apellidos = np.random.choice(apellidos, num_empleados)

# Generación de correos electrónicos únicos
emp_emails = [
    f"{nom.lower()}.{ape.lower()}{i}@empresa.com"
    for i, (nom, ape) in enumerate(zip(emp_nombres, emp_apellidos), start=1)
]

# Asignación de departamentos y puestos aleatorios
emp_dept_ids = np.random.choice(df_departamentos["department_id"], num_empleados)
emp_job_ids = np.random.choice(df_puestos["job_id"], num_empleados)

# Fechas de contratación aleatorias entre 2018 y 2025
start_date = datetime(2018, 1, 1)
end_date = datetime(2025, 12, 31)
time_between_dates = end_date - start_date
days_between_dates = time_between_dates.days

random_number_of_days = np.random.randint(0, days_between_dates, num_empleados)
emp_hire_dates = [start_date + timedelta(days=int(d)) for d in random_number_of_days]

# Generación de salarios dentro del rango de cada puesto
job_min_map = dict(zip(df_puestos["job_id"], df_puestos["min_salary"]))
job_max_map = dict(zip(df_puestos["job_id"], df_puestos["max_salary"]))

emp_salarios = [
    np.round(np.random.uniform(job_min_map[jid], job_max_map[jid]), 2)
    for jid in emp_job_ids
]

# Crear el DataFrame final de empleados
df_empleados = pd.DataFrame({
    "employee_id": emp_id,
    "first_name": emp_nombres,
    "last_name": emp_apellidos,
    "email": emp_emails,
    "hire_date": emp_hire_dates,
    "department_id": emp_dept_ids,
    "job_id": emp_job_ids,
    "salary": emp_salarios
})

# ==========================================
# Visualización previa
# ==========================================
print("--- DEPARTAMENTOS ---")
print(df_departamentos.head(2))
print("\n--- PUESTOS ---")
print(df_puestos.head(2))
print("\n--- EMPLEADOS (Muestra) ---")
print(df_empleados.head(3))
```

```python
import os
from pathlib import Path
import pandas as pd

# 1. Asegurar directorio base de forma segura (Se deja compatibilidad con Colab y otros entornos)
base_path = Path.cwd() if os.path.exists(os.getcwd()) else Path("/content")
DATA_DIR = base_path / "data" / "processed"

# Se crean las carpetas
DATA_DIR.mkdir(parents=True, exist_ok=True)

# Se definen las rutas de archivos para el dataframe empleados
path_csv = DATA_DIR / "employees.csv"
path_parquet = DATA_DIR / "employees.parquet"

# 2. Exportación de tablas a CSV y Parquet

# Se exporta dataframe empleados
df_empleados.to_csv(path_csv, index=False)
df_empleados.to_parquet(path_parquet, index=False, engine="pyarrow")

# Se exportan dataframes jobs y departments
df_puestos.to_csv(DATA_DIR / "jobs.csv", index=False)
df_puestos.to_parquet(DATA_DIR / "jobs.parquet", index=False)

df_departamentos.to_csv(DATA_DIR / "departments.csv", index=False)
df_departamentos.to_parquet(DATA_DIR / "departments.parquet", index=False)

print(f"✅ Archivos exportados con éxito en la carpeta: '{DATA_DIR}'\n")

# 3. Benchmark de Rendimiento: Cálculo de Tamaños y Compresión
size_csv_kb = path_csv.stat().st_size / 1024
size_parquet_kb = path_parquet.stat().st_size / 1024
compresion_pct = (1 - (size_parquet_kb / size_csv_kb)) * 100

# Presentación de la tabla comparativa
df_benchmark = pd.DataFrame(
    [
        {
            "Formato": "CSV (Texto plano)",
            "Tamaño (KB)": f"{size_csv_kb:.2f} KB",
            "Compresión": "Sin compresión",
            "Preserva Datatypes": "No (requiere casting al leer)",
        },
        {
            "Formato": "Parquet (Columnar / Snappy)",
            "Tamaño (KB)": f"{size_parquet_kb:.2f} KB",
            "Compresión": f"-{compresion_pct:.1f}%",
            "Preserva Datatypes": "Sí (metadata nativa)",
        },
    ]
)

print("--- 📉 TABLA COMPARATIVA DE ALMACENAMIENTO (BENCHMARK) ---")
display(df_benchmark)
```
